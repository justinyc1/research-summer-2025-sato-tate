# $p^{2}=49$

```sage
exponents = [a for a in range(1,25)]
endo = [f"-Z^{x}" for x in exponents]

print(exponents)
print("Endomorphism,", len(exponents))
print(", ".join(endo))
```

```sage
print(IntegerModRing(49).unit_gens())
```

```sage
# Group-acted exponents

newExponents = [(3*a % 49) for a in exponents]

# print("after Galois action, before reduction", newExponents)

groupactEndo = []
for a in newExponents:
    if a in exponents:
        groupactEndo.append(a)
    else:
        n = a - 49
        groupactEndo.append(n)
        
print(groupactEndo)

for j in groupactEndo:
    all_elms = True
    if abs(j) in exponents:
        pass
    else:
        print("Missing Z")
        all_elms = False
        
if all_elms:
    print("All elements preserved under group action")
```

```sage
# Getting gamma
# Simply track which elements conjugated, which didn't, and their indicies

for i, a in enumerate(exponents):
    for j, b in enumerate(groupactEndo):
        if a == b:
            print(f"{j, i} = I")
        elif a == abs(b):
            print(f"{j, i} = J")
```
